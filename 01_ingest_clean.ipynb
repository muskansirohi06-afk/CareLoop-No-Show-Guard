```python

import pandas as pd

# Load the original appointments dataset
df = pd.read_csv("data/appointments.csv")

# Check original data
print("Original shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

# Remove exact duplicate rows
df = df.drop_duplicates()

print("\nShape after removing duplicates:", df.shape)

# Clean booking channel names
print("\nBooking channels before cleaning:")
print(df["booking_channel"].value_counts(dropna=False))

df["booking_channel"] = (
    df["booking_channel"].str.strip().str.title()
)

print("\nBooking channels after cleaning:")
print(df["booking_channel"].value_counts(dropna=False))

# Convert date columns
df["appt_date"] = pd.to_datetime(
    df["appt_date"], errors="coerce"
)

df["booked_on"] = pd.to_datetime(
    df["booked_on"], errors="coerce"
)

# Remove rows with missing appointment date or status
df = df.dropna(subset=["appt_date", "status"])

# Final checks
print("\nFinal shape:", df.shape)

print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nDuplicate rows after cleaning:", df.duplicated().sum())
```

```python

import numpy as np

# 1. Handle invalid ages
invalid_age = (df["age"] < 1) | (df["age"] > 100)

print("Invalid ages:", invalid_age.sum())

df.loc[invalid_age, "age"] = np.nan

# Fill missing ages with median
age_median = df["age"].median()
df["age"] = df["age"].fillna(age_median).round().astype(int)

print("Missing ages after filling:", df["age"].isnull().sum())
print("Age median used:", age_median)

# 2. Fill missing distances with median
distance_median = df["distance_km"].median()
df["distance_km"] = df["distance_km"].fillna(distance_median)

print("\nDistance median used:", distance_median)

# 3. Calculate distance outlier limits using IQR
Q1_distance = df["distance_km"].quantile(0.25)
Q3_distance = df["distance_km"].quantile(0.75)
IQR_distance = Q3_distance - Q1_distance

lower_distance = Q1_distance - 1.5 * IQR_distance
upper_distance = Q3_distance + 1.5 * IQR_distance

print("\nDistance Q1:", Q1_distance)
print("Distance Q3:", Q3_distance)
print("Distance IQR:", IQR_distance)
print("Distance lower limit:", lower_distance)
print("Distance upper limit:", upper_distance)

distance_outliers = (
    (df["distance_km"] < lower_distance) |
    (df["distance_km"] > upper_distance)
).sum()

df["distance_km"] = df["distance_km"].clip(
    lower=lower_distance, upper=upper_distance
)

print("Distance outliers capped:", distance_outliers)

# 4. Calculate fee outlier limits using IQR
Q1_fee = df["fee_inr"].quantile(0.25)
Q3_fee = df["fee_inr"].quantile(0.75)
IQR_fee = Q3_fee - Q1_fee

lower_fee = Q1_fee - 1.5 * IQR_fee
upper_fee = Q3_fee + 1.5 * IQR_fee

print("\nFee Q1:", Q1_fee)
print("Fee Q3:", Q3_fee)
print("Fee IQR:", IQR_fee)
print("Fee lower limit:", lower_fee)
print("Fee upper limit:", upper_fee)

fee_outliers = (
    (df["fee_inr"] < lower_fee) |
    (df["fee_inr"] > upper_fee)
).sum()

df["fee_inr"] = df["fee_inr"].clip(
    lower=lower_fee, upper=upper_fee
)

print("Fee outliers capped:", fee_outliers)

# 5. Round values
df["distance_km"] = df["distance_km"].round(1)
df["fee_inr"] = df["fee_inr"].round().astype(int)

print("\nFinal missing values:")
print(df[["age", "distance_km", "fee_inr"]].isnull().sum())

print("\nFinal maximum distance:", df["distance_km"].max())
print("Final maximum fee:", df["fee_inr"].max())
```

```python

import numpy as np

# 1. Calculate lead days
df["lead_days"] = (
    df["appt_date"] - df["booked_on"]
).dt.days

# 2. Create no_show column
df["no_show"] = (df["status"] == "No-Show").astype(int)

# 3. Display basic information
print("Lead days minimum:", df["lead_days"].min())
print("Lead days maximum:", df["lead_days"].max())

print("\nNo-show counts:")
print(df["no_show"].value_counts())

print("\nNo-show percentage:",
      round(df["no_show"].mean() * 100, 1), "%")

# 4. Calculate appointments booked more than 14 days in advance
more_than_14 = (df["lead_days"] > 14).sum()

print("\nAppointments with lead days > 14:", more_than_14)
print("Percentage:",
      round(more_than_14 / len(df) * 100, 1), "%")

# 5. Create NumPy feature array
X = df[[
    "age",
    "distance_km",
    "fee_inr",
    "lead_days"
]].to_numpy()

print("\nNumPy array shape:", X.shape)
print("Feature means:", np.round(X.mean(axis=0), 2))

# 6. Calculate distance z-scores
distance_z = (
    df["distance_km"] - df["distance_km"].mean()
) / df["distance_km"].std()

print("\nDistance values with absolute z-score > 2:",
      (np.abs(distance_z) > 2).sum())
```

```python

import pandas as pd
import requests

# Load the original clinic dataset
clinics_df = pd.read_csv("data/clinics.csv")

# Store temperatures
temperatures = []

# Open-Meteo API URL
url = "https://api.open-meteo.com/v1/forecast"

# Process each clinic
for _, clinic in clinics_df.iterrows():

    temp_c = None
    reason = None

    # Try twice if the API fails
    for attempt in range(2):
        try:
            params = {
                "latitude": clinic["latitude"],
                "longitude": clinic["longitude"],
                "hourly": "temperature_2m",
                "forecast_days": 1
            }

            response = requests.get(
                url, params=params, timeout=10
            )

            if response.status_code != 200:
                reason = f"HTTP status {response.status_code}"
                continue

            weather_data = response.json()

            hourly = weather_data.get("hourly")
            if not hourly or not hourly.get("temperature_2m"):
                reason = "Hourly temperature data missing"
                continue

            temp_c = hourly["temperature_2m"][0]
            reason = None
            break

        except requests.exceptions.RequestException as e:
            reason = str(e)

        except (ValueError, KeyError, TypeError, IndexError) as e:
            reason = str(e)

    temperatures.append(temp_c)

    if temp_c is None:
        print("API failed for", clinic["clinic_name"], "-", reason)
    else:
        print(clinic["clinic_name"], "Temperature:", temp_c)

# Add temperature column
clinics_df["temp_c"] = temperatures

# Save enriched dataset
clinics_df.to_csv("clinics_enriched.csv", index=False)

print("\nEnriched clinic data:")
display(clinics_df)

print("\nFile saved: clinics_enriched.csv")
print("Total clinics:", len(clinics_df))
print("Missing temperatures:", clinics_df["temp_c"].isnull().sum())
```

```python

# TASK 1.6 — EXPORT CLEAN DATA AND CREATE SQL LOAD SCRIPT

import pandas as pd
import numpy as np

# 1. Use the cleaned appointment dataframe
cleaned_df = df.copy()

# 2. Required 16 columns in the exact order
required_columns = [
    "appointment_id",
    "patient_id",
    "clinic_id",
    "booked_on",
    "appt_date",
    "age",
    "gender",
    "distance_km",
    "fee_inr",
    "booking_channel",
    "visit_type",
    "reminder_sent",
    "prior_no_shows",
    "status",
    "lead_days",
    "no_show"
]

# Check that all required columns exist
missing_columns = [
    col for col in required_columns if col not in cleaned_df.columns
]
assert not missing_columns, f"Missing columns: {missing_columns}"

# 3. Keep only required columns in the correct order
cleaned_df = cleaned_df[required_columns].copy()

# 4. Format both date columns as YYYY-MM-DD
for col in ["booked_on", "appt_date"]:
    cleaned_df[col] = pd.to_datetime(
        cleaned_df[col], dayfirst=True, errors="raise"
    ).dt.strftime("%Y-%m-%d")

# 5. Sort by appointment_id
cleaned_df = cleaned_df.sort_values(
    "appointment_id"
).reset_index(drop=True)

# 6. Validate the data before exporting
assert cleaned_df["appointment_id"].is_unique, \
    "Duplicate appointment IDs found!"

assert not cleaned_df.isnull().any().any(), \
    "Missing values found in appointment data!"

print("Unique appointment IDs:", cleaned_df["appointment_id"].nunique())
print("Missing values:", cleaned_df.isnull().sum().sum())
print("Final appointment shape:", cleaned_df.shape)

# 7. Save appointments_clean.csv
cleaned_df.to_csv("appointments_clean.csv", index=False)

# 8. Load enriched clinic data
clinics_df = pd.read_csv("clinics_enriched.csv")

# Ensure all required clinic columns are present
clinic_columns = [
    "clinic_id", "clinic_name", "city", "tier",
    "latitude", "longitude", "temp_c"
]

assert all(col in clinics_df.columns for col in clinic_columns)

clinics_df = clinics_df[clinic_columns].copy()

# 9. Function to convert Python values into SQL values
def sql_value(value):
    if pd.isna(value):
        return "NULL"

    if isinstance(value, (int, np.integer)):
        return str(int(value))

    if isinstance(value, (float, np.floating)):
        return str(float(value))

    value = str(value).replace("'", "''")
    return "'" + value + "'"

# 10. Create load_data.sql
with open("load_data.sql", "w", encoding="utf-8") as file:

    # Insert clinics FIRST (foreign key requirement)
    for _, row in clinics_df.iterrows():
        values = [
            sql_value(row[col]) for col in clinic_columns
        ]

        file.write(
            "INSERT INTO clinics VALUES ("
            + ",".join(values)
            + ");\n"
        )

    # Insert appointments AFTER clinics
    for _, row in cleaned_df.iterrows():
        values = [
            sql_value(row[col]) for col in required_columns
        ]

        file.write(
            "INSERT INTO appointments VALUES ("
            + ",".join(values)
            + ");\n"
        )

# 11. Final verification
with open("load_data.sql", "r", encoding="utf-8") as file:
    sql_lines = [
        line for line in file
        if line.strip().upper().startswith("INSERT INTO")
    ]

clinic_inserts = sum(
    line.startswith("INSERT INTO clinics") for line in sql_lines
)

appointment_inserts = sum(
    line.startswith("INSERT INTO appointments") for line in sql_lines
)

print("\nTASK 1.6 COMPLETED")
print("Appointments exported:", len(cleaned_df))
print("Clinics exported:", len(clinics_df))
print("Clinic INSERT statements:", clinic_inserts)
print("Appointment INSERT statements:", appointment_inserts)
print("Total INSERT statements:", len(sql_lines))
print("Files saved: appointments_clean.csv and load_data.sql")
```
